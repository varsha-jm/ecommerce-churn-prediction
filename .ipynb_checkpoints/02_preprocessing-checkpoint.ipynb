# E-Commerce Churn — Preprocessing & Feature Engineering

## Steps completed:
- Loaded raw e-commerce churn dataset
- Filled missing numeric values using median imputation
- Engineered four new churn-related features:
  - is_new_customer
  - is_inactive
  - is_high_cashback
  - high_risk
- Encoded all categorical columns using LabelEncoder
- Split dataset into X (features) and y (target)
- Performed stratified train-test split
- Applied StandardScaler for feature normalization
- Saved processed dataset for model training

```python
df.isnull().sum()
```

```python
num_cols=df.select_dtypes(include='number').columns
num_cols
```

```python
df[num_cols]=df[num_cols].fillna(df[num_cols].median())
```

```python
df.isnull().sum()
```

```python
df['is_new_customer']=(df['Tenure'] < 3).astype(int)
```

```python
df[['Tenure','is_new_customer']].head(10)
```

```python
df['is_inactive']=(df['DaySinceLastOrder'] > 20).astype(int)
```

```python
df[['DaySinceLastOrder','is_inactive']].head(10)
```

```python
df.groupby('is_inactive')['Churn'].mean()
```

```python
df['is_high_cashback']=(df['CashbackAmount']>200).astype(int)
```

```python
df[['CashbackAmount','is_high_cashback']].head(10)
```

```python
df.groupby('is_high_cashback')['Churn'].mean()
```

```python
df['high_risk'] = ((df['Complain'] == 1) & (df['DaySinceLastOrder'] > 15)).astype(int)
print("created")
```

```python
'high_risk' in df.columns
```

```python
df.columns.tolist()[-5:]
```

```python
df.drop('high-risk', axis=1, inplace=True)
```

```python
df.columns.tolist()[-5:]
```

```python
cat_cols=df.select_dtypes(include='object').columns
cat_cols
```

```python
le=LabelEncoder()
for col in cat_cols:
    df[col]=le.fit_transform(df[col])
```

```python
X=df.drop('Churn',axis=1)
y=df['Churn']
```
